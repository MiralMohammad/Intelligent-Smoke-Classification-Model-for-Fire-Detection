# **Training**

```python
from sklearn.model_selection import train_test_split

x=df.drop('Fire Alarm',axis=1)
y=df['Fire Alarm']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=42)
```

### SMOTE Oversampling

Before training, we want to solve the class imbalance problem. I’m going to Use SMOTE method, and then evaluate it along with the original data to see which one gives the best results.

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)
```

```python
#scaler
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_train_smote=scaler.transform(x_train_smote)
x_test=scaler.transform(x_test)
```

### Model before tuning

```python
from sklearn.ensemble import AdaBoostClassifier
model_pre_tuning=AdaBoostClassifier(random_state=42)
model_pre_tuning.fit(x_train,y_train)
y_pred_pre_tuning=model_pre_tuning.predict(x_test)
```

```python
model_smote_pre_tuning=AdaBoostClassifier(random_state=42)
model_smote_pre_tuning.fit(x_train_smote,y_train_smote)
y_pred_smote_pre_tuning=model_smote_pre_tuning.predict(x_test)
```

### Hyper parameter tuning

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import RandomizedSearchCV

base_tree = DecisionTreeClassifier(random_state=42)

adaboost = AdaBoostClassifier(
    estimator=base_tree,
    random_state=42
)


param_grid = {
    "n_estimators": [50, 100, 200],
    "learning_rate": [0.001, 0.01, 0.1],
    "estimator__max_depth": [1, 2]
}

random_search = RandomizedSearchCV(
    estimator=adaboost,
    param_distributions=param_grid,
    n_iter=10,       # was 30
    scoring="precision",
    cv=3,           # was 5
    n_jobs=-1,
    random_state=42,
    verbose=1
)

random_search.fit(x_train, y_train)

print("Best params:", random_search.best_params_)
print("Best CV Score:", random_search.best_score_)
```

Training the model on the original data:

```python
from sklearn.tree import DecisionTreeClassifier
base_tree = DecisionTreeClassifier(max_depth=1, random_state=42)
model=AdaBoostClassifier(estimator=base_tree,n_estimators=200,learning_rate=0.01,random_state=42)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

Training the model on the SMOTE copy:

```python
model3=AdaBoostClassifier(estimator=base_tree,n_estimators=200,learning_rate=0.01,random_state=42)
model3.fit(x_train_smote,y_train_smote)
y_pred_smote=model3.predict(x_test)
```

# Evaluation

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

## **Baseline**

Dump Baseline model that predicts fire in all cases, achieving perfect recall but showing no real learning or decision ability.

```python
from sklearn.dummy import DummyClassifier

dummy = DummyClassifier(random_state=42)
dummy.fit(x_train, y_train)
y_pred_dummy = dummy.predict(x_test)


print("DUMMY BASELINE:")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_dummy))
print("Accuracy : ",accuracy_score(y_test, y_pred_dummy))
print("F1-score : ",f1_score(y_test, y_pred_dummy))
print("Precision: ",precision_score(y_test, y_pred_dummy))
print("Recall   : ",recall_score(y_test, y_pred_dummy))
```

Logistic Regression Baseline

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(random_state=42)
lr.fit(x_train, y_train)

y_pred_lr = lr.predict(x_test)

print("Logistic Regression Results")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_lr))
print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("F1-score:", f1_score(y_test, y_pred_lr))
print("Precision: ",precision_score(y_test, y_pred_lr))
print("Recall   : ",recall_score(y_test, y_pred_lr))
```

Compute learning curve to analyze model performance as training size increases

```python
from sklearn.model_selection import learning_curve
import numpy as np
train_sizes, train_scores, test_scores = learning_curve(
    estimator=model,
    X=x_train,
    y=y_train,
    cv=5,
    scoring="f1",
    n_jobs=-1,
    train_sizes=np.linspace(0.1, 1.0, 5)
)

train_mean = np.mean(train_scores, axis=1)
test_mean = np.mean(test_scores, axis=1)
train_mean, test_mean
```

Plot learning curve for visual comparison between training and validation scores

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 5))
plt.plot(train_sizes, train_mean, marker='o', label='Train F1-score')
plt.plot(train_sizes, test_mean, marker='s', label='Validation F1-score')
plt.title('Learning Curve - AdaBoost (Original data)')
plt.xlabel('Training Set Size')
plt.ylabel('F1-score')
plt.legend()
plt.grid(True)
plt.show()
```

**Comparison between baseline models and the tuned model**

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models_preds = [
    ("Dummy Baseline", y_pred_dummy),
    ("Naive Bayes Baseline", y_pred_nb),
    ("Logistic Regression", y_pred_lr),
    ("AdaBoost (Tuned)", y_pred)
]

results = []

for name, y_pred_model in models_preds:
    acc = accuracy_score(y_test, y_pred_model)
    prec = precision_score(y_test, y_pred_model)
    rec = recall_score(y_test, y_pred_model)
    f1 = f1_score(y_test, y_pred_model)
    results.append({
        "Model": name,
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1-score": f1
    })

df_results = pd.DataFrame(results)
print(df_results)
```
